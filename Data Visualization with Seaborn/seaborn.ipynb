Data visualization with Seaborn

Seaborn is a Python visualizaiton library based on matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics .Seaborn helps in creating complex visualization with just a few lines fo code. In this lesson, we will cover the basics of seaborn , including creating various types of plots and customizing them.

```python
import seaborn as sns
```

```python
##basic plotting with seaborn 
tips=sns.load_dataset('tips')
tips
```

```python
## create a scatter plot 
import matplotlib.pyplot as plt

sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("Scatter plot of total bill vs tip")
plt.show()
```

```python
## line plot 
sns.lineplot(x='size',y='total_bill',data=tips)
plt.title("line  plot of total bill by size")
plt.show()
```

```python
## Cetegorical  plots
## bar plot
sns.barplot(x='day',y='total_bill',data=tips)
plt.title("bar plot of total bill by day")
plt.show()
```

```python
## box plot 
sns.boxplot(x="day",y='total_bill',data=tips)
```

```python
## violin plot 
sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
### histogram
sns.histplot(tips['total_bill'],bins=10,kde=True)
```

```python
## KDE plot
sns.kdeplot(tips['total_bill'],fill='True')
```

```python
#Pairplot
sns.pairplot(tips)
```

```python
## heatmap
corr=tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python
import pandas as pd
sales_df=pd.read_csv('Online_Sales_Data.csv')
sales_df.head()
```

```python
##plot total sales by product
plt.figure(figsize=(10,6))
sns.barplot(x='Product Category',y="Total Revenue",data=sales_df,estimator=sum)
plt.xlabel('product')
plt.ylabel('total Sales')
plt.show()
```

```python
##plot total sales by Region
plt.figure(figsize=(10,6))
sns.barplot(x='Region',y="Total Revenue",data=sales_df,estimator=sum)
plt.title('total sales by region')
plt.xlabel('Region')
plt.ylabel('total Sales')
plt.show()
```
