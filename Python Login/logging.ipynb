Logging is a crucial aspect of any application, providing a way to track events, errors, and operational information. Python's built-in logging module offers a flexible framework for emitting log messages from python programs.In this lesson, we will cover the  basics of logging, including how to configure logging,log levels,and best practices for using logging in python applications.

```python
import logging
```

```python
## configure the basic logging settings 
logging.basicConfig(level=logging.DEBUG)

## LOG MESSAGES WITH  DIFFERENT  SEVERITY LEVELS 
logging.debug("this is a debug message")
logging.info("this is a debug message")
logging.warning("this is a debug message")
logging.error("this is a debug message")
logging.critical("this is a debug message")
```

LOG LEVELS

PYTHON'S LOGGING MODULE HAS SEVERAL LOG LEVELS INDICATING THE SEVERITY  OF EVENTS . THE DEFAULT LEVELS ARE:

DEBUG:DETAILED INFORMATION,TYPICALLY OF INTEREST ONLY WHEN DIAGNOSIN PROBLEM.

Info :confirmation that things are working as expected . 
Warning: An indication that something unexpected or indicative of some problem in the near future (eg. 'disk space low) . the software is still working as expected. 

error: due to a more seriou problem , the software has not been able to perform some function.

critical: A very serious error, indicating thatd the program itself may be unable to continue running.

```python
## configure logging
logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datafmt='%Y-%m-%d %H:%M:%S'
)

## LOG MESSAGES WITH  DIFFERENT  SEVERITY LEVELS 
logging.debug("this is a debug message")
logging.info("this is a debug message")
logging.warning("this is a debug message")
logging.error("this is a debug message")
logging.critical("this is a debug message")
```
