```python
## Dictionaries are  unordered collection of items. They store data in key -value pairs. Keys  must be unique and immutable (eg strings numbers or tuples) while values can be of any type.
```

```python
## creating Dictionaries 
empty_dict ={}
print(type(empty_dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
student={"name" : "krish","age":32}
print(student)
print(type(student))
```

```python
## error
student={"name":"krish","age":32,"name":24}
print(student)## single key is always used 

```

```python
## accessing dictionary elements 
print(student["name"])

## accessing using get() method 
print(student.get('grade'))
```

```python
## modifying dictionary elements 
## dictionary are mutable , so u can add , update or  delete elements 
print(student)
```

```python
student["age"]=33
print(student)
student["address"]="India"
print(student)
```

```python
del student['age']  ## delete key and value pair 

print(student)
```

```python
## dictionary methods 

keys=student.keys()
print(keys)
values=student.values()
print(values)


items=student.items() ## all key value pair 
print(items)
```

```python
## shallow copy 
student_copy=student
print(student)
print(student_copy)
```

```python
student["name"]="krish2"
print(student)
print(student_copy)
```

```python
student_copy1=student.copy() ## shallow copy 
print(student_copy)
print(student)
```

```python
student["name"]="Krish3"
print(student_copy1)
print(student)
```

```python
### iterating over dictionaries 
## you can use loops to iterate over dictionarie, keys values or items 

## iterating over keys 
for keys  in student.keys():
    print(keys)
```

```python
## iterate over values 
for value  in student.values():
    print(value)
```

```python
## iterates over key value pairs 
for key,value in student.items():
    print(f"{key}:{value}")
```

```python
## nested dictionaries 
students={
    "student1":{"name":"krish","age":32},
    "student2":{"name":"ram","age":33},
}
print(students)
```

```python
## access nested dictionaries  elements 
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
## iterating over nested dictionaries 
for student_id, student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## dictionary comprehension 

squares={ x:x**2 for x in range(5)}
print(squares)
```

```python
## condition dictionary comprehension 
events={x:x**2 for x in range(10) if x%2==0}
print(events)
```

```python
## practical examples 

## use a dictionary to count the frequency of elemets in list 

numbers =[1,2,2,3,3,3,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else: 
        frequency[number]=1
        
print(frequency)
```

```python
## merge 2 dictinaries into one 

dict1={"a":1,"b":2}
dict2={"b":3,"c":4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```
