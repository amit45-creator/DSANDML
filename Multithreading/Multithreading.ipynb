```python
### Multithreading 
## when to use multithreading 
### I/o -bound tasks: task that spend more than time waiting for I/O operattions (eg file operation ,network request)
### concurrent execution; when you want to improve the throughput of your applications by performing multiple operations concurrently.

import threading 
import time

def print_number():
    for i in range(5):
      time.sleep(2)
      print(f"Number:{i}")
      
def print_letters():
    for letters in "abcde":
        time.sleep(2)
        print(f"Letter:{letters}")  
        
##create 2 threads
t1=threading.Thread(target=print_number)
t2=threading.Thread(target=print_letters)

         
t=time.time() 
##start the thread

t1.start()
t2.start()

### wait for the threads to complete
t1.join()
t2.join()
     
# print_number()
# print_letters()
finished_time=time.time()-t
print(finished_time)
```
